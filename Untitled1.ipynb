```python


import pandas as pd
import sqlite3
from pathlib import Path
```

```python
df = pd.read_excel("FIFA_World_Cup_2026_Player_Data.xlsx")
print("Rows:", len(df))
print("Columns:", len(df.columns))
df.head()
```

```python
conn = sqlite3.connect("fifa_wc2026.db")

df.to_sql(
    "players",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM players
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    position,
    COUNT(*) AS player_count
FROM players
GROUP BY position
ORDER BY player_count DESC;
"""

position_summary = pd.read_sql_query(query, conn)

position_summary
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Remove rows with missing Age or Rating
plot_df = df[
    (df["age"] > 0) &
    (df["player_rating"] > 0)
][["age", "player_rating"]].dropna()

# Scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(
    plot_df["age"],
    plot_df["player_rating"],
    alpha=0.5,
    color="steelblue"
)

# Calculate trend line
z = np.polyfit(plot_df["age"], plot_df["player_rating"], 1)
trend = np.poly1d(z)

# Draw trend line
plt.plot(
    plot_df["age"],
    trend(plot_df["age"]),
    color="red",
    linewidth=2,
    label="Trend line"
)

plt.title("Player Age vs Rating")
plt.xlabel("Age")
plt.ylabel("Player Rating")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Remove missing and zero values
plot_df = df[
    (df["market_value_eur"] > 10000000) &
    (df["player_rating"] > 0)
][["market_value_eur", "player_rating"]].dropna()

# Scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    plot_df["market_value_eur"],
    plot_df["player_rating"],
    alpha=0.5,
    color="steelblue"
)

# Trend line
z = np.polyfit(
    plot_df["market_value_eur"],
    plot_df["player_rating"],
    1
)

trend = np.poly1d(z)

plt.plot(
    plot_df["market_value_eur"],
    trend(plot_df["market_value_eur"]),
    color="red",
    linewidth=2,
    label="Trend line"
)

# Labels and formatting
plt.title("Market Value vs Player Rating")
plt.xlabel("Market Value (€)")
plt.ylabel("Player Rating")
plt.xscale("log")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

top_awards = df.nlargest(10, "player_of_match_awards")

plt.figure(figsize=(10, 6))

plt.barh(
    top_awards["player_name"],
    top_awards["player_of_match_awards"],
    color="steelblue"
)

plt.title("Top 10 Players by Number of Awards")
plt.xlabel("Number of Awards")
plt.ylabel("Player")

# Add nationality to player names
labels = [
    f"{player} ({nationality})"
    for player, nationality in zip(
        top_awards["player_name"],
        top_awards["team"]
    )
]

plt.yticks(range(len(top_awards)), labels)

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

top_goals = df.nlargest(10, "total_goals_tournament")

plt.figure(figsize=(10, 6))

plt.barh(
    top_goals["player_name"],
    top_goals["total_goals_tournament"],
    color="steelblue"
)

plt.title("Top 10 Players by Total Tournament Goals")
plt.xlabel("Total Goals")
plt.ylabel("Player")

# Add nationality
labels = [
    f"{player} ({nationality})"
    for player, nationality in zip(
        top_goals["player_name"],
        top_goals["nationality"]
    )
]

plt.yticks(range(len(top_goals)), labels)

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Remove missing and zero values
age_data = df[df["age"] > 0]["age"].dropna()

plt.figure(figsize=(10, 6))

plt.hist(
    age_data,
    bins=range(int(age_data.min()), int(age_data.max()) + 2),
    color="steelblue",
    edgecolor="black"
)

plt.title("Age Distribution of Players")
plt.xlabel("Age")
plt.ylabel("Number of Players")

plt.xticks(range(int(age_data.min()), int(age_data.max()) + 1, 2))

plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Remove missing and zero values
plot_df = df[
    (df["age"] > 0) &
    (df["position"].notna())
]

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=plot_df,
    x="position",
    y="age",
    color="steelblue"
)

plt.title("Age Distribution by Player Position")
plt.xlabel("Position")
plt.ylabel("Age")

plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
metrics = [
    "player_rating",
    "performance_score",
    "offensive_contribution",
    "defensive_contribution",
    "possession_impact",
    "pressure_resistance",
    "creativity_score",
    "consistency_score",
    "clutch_performance_score",
    "goals",
    "assists",
    "shots_on_target",
    "expected_goals_xg",
    "expected_assists_xa",
    "key_passes",
    "pass_accuracy",
    "successful_dribbles",
    "tackles",
    "interceptions",
    "recoveries",
    "minutes_played",
    "market_value_eur"
]

import matplotlib.pyplot as plt
import seaborn as sns

corr = df[metrics].corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Correlation Between Player Rating and Key Performance Metrics")
plt.tight_layout()
plt.show()
```

```python
correlations = (
    df[metrics]
    .corr()["player_rating"]
    .drop("player_rating")
    .drop("performance_score")
    .sort_values(ascending=False)
)

print(correlations)
```
